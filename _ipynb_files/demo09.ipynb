### dictionary comprehension (and use-case)

```python
import json

# JSON string (keys are strings)
s = '{"1": "apple", "2": "banana"}'

# Load JSON
data = json.loads(s)
print(f"before: {data}")
# Convert keys to integers: **with a dictionary comprehension**
data = {int(k): v for k, v in data.items()}

print(f"after: {data}")
```
