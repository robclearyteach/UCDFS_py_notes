# Catching exceptions

Unhandled exceptions force the program to stop running (and entirely exit if we're working in a non-interactive environment rather than something like jupyter). We usually want to avoid this.

The way to work with exceptions, and have our programmes deal with the issues, is to "catch" them whenever we expect them and have a way of handling them.
(We'll later talk about why we should not catch exceptions that we cannot handle)

To do so, we use a `try` statement with an `except` clause:

```python
# From the exception class hierarchy above, you might have noticed that
# all (regular) exceptions inherit from the class `Exception`

#You might create a catch all instead of worrying about particular exception types, you could just handle `Exception` like this:
age = 21
response = input("Please enter how many chunks of your life you'd like to examine: ")

try:
    chunks = int(response)
    raise MemoryError()  #simulate unexpected exception occurs
    print(f"Let's now examine your life as composed of {chunks} chunks of {age/chunks:0.1f} years each)")
except Exception as e:
    print( "the exception object: ", e )
    print(f"Your input of '{response}' is invalid - you must enter a positive integer")

print("program continues...")
```

The issue is that by catching `Exception`, we might catch many other types of exceptions that we don't know how to handle, and remove the ability to see what happened - this is also known as "swallowing exceptions".

A common example is of inadvertantly catching `MemoryError` which occurs when the computer runs low on memory. Some exception classes like `MemoryError` cannot typically be handled by our code, but can be useful to see when troubleshooting programmes, but if we catch it and fail to handle it, then we wouldn't be able to easily identify that our programme is accidentally eating up the computer's memory.

Try to avoid using the general `Exception` unless you're absolutely sure that you want to handle all possible types of exceptions in the exact same manner.

```python
# Going back to our age example:

try: 
    response = input("Please enter your age: ")
    age = int(response)
    print(f"Next year you'll be {age + 1} years old")  # If an exception was raised, this would not run
except ValueError:  # We go into this block if the try block raised an exception
    print(f"Your response '{response}' was not a valid number")

print(f"program continues: age does not exist: {age}")

#Is there a difference if we replace our ValueError with a catch all exception? 
```

```python
# Or even better, we can use this in a loop, to keep asking for input until we get a valid value:
response = None
while True:
    try: 
        response = int(input("Please enter your age: "))
        break  # Only if the line above didn't raise an exception, we break the loop
    except ValueError:
        if response is not None:
            print(f"Your response '{response}' was not a valid number")
        else:
            print(f"Your entered age was not valid")
            
#How about now what if replace our exception with a catch all?
            
# Now we know that `age` is a valid integer
print(f"Next year you'll be {age + 1} years old")  # If an exception was reached, this would not run
```

```python
# Sometimes the same part of the code can cause different types of exceptions
# We can add multiple except clauses to the same try statement, to catch and handle each type:

age = 21
response = input("Please enter how many chunks of your life you'd like to examine: ")

try:
    chunks = int(response)
    print(f"Let's now examine your life as composed of {chunks} chunks of {age/chunks:0.1f} years each)")
except ValueError:
    print("error")
except ZeroDivisionError:
    print("The number of chunks cannot be zero")
    
print("program continues...")
```

# Else and Finally
These are two additional optional (and rarely used) clauses we can use when catching exceptions:

* `else` - the block in this clause is only run if no exception was raised in the `try` block - it's the same as putting the code at the end of the try block, but it specifies that we don't want to catch exceptions here
* `finally` - the block in this clause is run regardless of whether there was an exception or not - it's sometimes used for cleanup operations that should be done in any case (e.g. releasing a resource)

```python
age = 21
response = input("Please enter how many chunks of your life you'd like to examine: ")

try:
    chunks = int(response)
except (ValueError, ZeroDivisionError):
    print(f"Your input of '{response}' is invalid - you must enter a positive integer")
else: 
    #avoid ... really avoid...
    try:
        print(f"Let's now examine your life as composed of {chunks} chunks of {age/chunks:0.1f} years each)")
    except ZeroDivisionError:
        print("ok, catching in the else but boy i shouldn't do that!")
finally:
    print("Thank you for using our programme, have a nice day")
```

```python
# `finally` is run even for exceptions that we didn't catch
# so we can use try...finally blocks, without any except clause, just for cleanup
try:
    print("Opening some external connection")
    print("Doing some work")
    raise RuntimeError("Oops, we did something bad and have to abort")
finally:
    print("Before aborting, we close the external connection, to allow others to use it")
    # The above will be printed after the exception is raised but before it is processed
    
#What happens if we add an else here can you notice the difference?
```

# Exception objects and special exception classes
Being classes, various types of exceptions offer some additional functionality, which we can access by using the exception object via the `as` keyword.

```python
# For example, let's try to read from an non-existing file:
try:
    f = open("no_such_file")
except OSError as e:
    # We can print the exception object itself (using its __str__ method) to see some general detail
    print(e)
```

```python
try:
    f = open("no_such_file")
except OSError as e:
    # Knowing that it's an IOError object, we can access some specific attributes of that class:
    print(f"{e.errno=}, {e.filename=}, {e.strerror=}")
```

# The `with` statement

Python actually offers us a nicer way to work with files without having to deal with closing them.

The `with` statement allows us to acquire a resource for the duration of a block, and it will be cleaned up automatically for us, regardless of whether an exception was raised (removing the need for a try block)

```python
with open("myfile.txt") as f:
    for line in f:
        print(">>!", line.rstrip() ) 
        
# with open("myfile.txt") as f:
#     print(f.read())
#     f.write("something")  # Will fail, because we opened it in read-only mode

# After the with block the file object is guaranteed to be closed properly,
# no matter what exceptions occurred within the block.

# But we would still use a try...except if we want to handle specific errors
```

Many classes representing external resources such as database connections support the `with` statement.
To have our own objects support `with` we just need to create `__enter__` and `__exit__` methods which would be run by Python at the entry and exit of the `with` block respectively (but this is beyond our scope)



# Exceptions in function calls

Exceptions stop the execution of the current function and if uncaught "propagate" upwards, to the function that called it.

This means that as part of the design of our code, we can decide at what level of our program we might want to catch exceptions. We would then typically try to have our function names reflect whether they are likely to fail.

```python
# For example, a function named calculate_average should not be expected to work on an empty collection
def calculate_average(comma_separated_nums):
    """Calculate average of nums in comma-separated-string.
    
    Raise ZeroDivisionError if no nums provided in `comma_separated_nums`
    """    
    comma_separated_nums = comma_separated_nums.split(",")
    print( comma_separated_nums )
    nums = []
    for item in comma_separated_nums:
        item = item.strip()
        if not item:
            continue
        nums.append(float(item))

    return sum(nums) / len(nums)    #ZeroDivision if len(nums) is zero


# We'll handle any input issues outside the above function
while True:
    try:
        response = input("Enter comma separated numbers, or 'done' to quit")
        if response.lower() == "done":
            break  # Stop when given empty input
        print("Avg of list: ", calculate_average(response))
    except ZeroDivisionError:
        print(f"Invalid input: '{response}', make sure there's at least one number")
```

# Custom exception classes
Sometimes, instead of using a built-in exception class in our code, we may want to create a custom one, that best describes our issue.
To do so, we just need to create a new class that inherits from `Exception`. Typically we don't need to add any new behaviour and can just put `pass` in it. 

```python
class PythonLessonException(Exception):
    pass


def teach(lesson):
    if lesson > 6:
        raise PythonLessonException("No can do, there are only 6 lessons in this course, get some good rest now")


teach(lesson=7)
```

```python
# And then we can catch just this custom exception type

try:
    teach(lesson=7)
except PythonLessonException:
    print("Well, we do have a tiny bit more to cover before we're done")
#Create your own exception 
```
